```python
import plotly.express as px

# 1. Platform wise Value - Bar Chart
fig1 = px.bar(df, x='Platform', y='Value_Std', color='Platform',
              title='Platform Wise Value_Std')
fig1.show()

# 2. Year wise Trend - Line Chart
fig2 = px.line(df, x='Year', y='Value_Std', color='Platform',
               title='Year Wise Trend - Including all Platforms')
fig2.show()

# 3. Metric wise - Pie Chart
fig3 = px.pie(df, names='Metric', values='Value_Std',
              title='Metric Wise Distribution')
fig3.show()
```
